```python
#Write a fn takes a no as argument and return odd or even
def even_odd(num):
  if num % 2 == 0:
    return num/2
  else:
    return 3*num +1
even_odd(77)
```

```python
#factorial of n
def fact(n):
  if n <=1 :
    return 1
  else:
    return n * fact(n-1)
fact(5)
```

```python
#sum of first n odd numbers
def sum(n):
  if n == 0:
    return 0
  else:
    return n ** 2
sum(5)
```

```python
def fibo(n):
  if n <= 1:
    return n
  else:
    return fibo(n-1) + fibo(n-2)
fibo(5)
```

```python
#Reverse a number
def rev(n):
  reev=""
  while n >= 1:
    reev += str(n % 10)
    n = n// 10
  return int(reev)
rev(123)
```

```python
#sum of digits
def sum_digits(n):
  summ=0
  while n > 0:
    summ += n % 10
    n= n//10
  return summ
sum_digits(123)
```

```python
#palindrome
def pali(n):
  reev=""
  num=n
  while n >= 1:
    reev += str(n % 10)
    n = n// 10
  if int(reev) == num:
    return True
  else:
    return False
pali(1232)
```
